```python

# =====================================================
# Task 1: Environment Setup
# =====================================================
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
 
print("=" * 60)
print("TASK 1: Environment Setup")
print("=" * 60)
print(f"Gymnasium version: {gym.__version__}")
print(f"NumPy version: {np.__version__}")
print("Successful installation and Gymnasium version displayed.\n")
```

```python

# =====================================================
# Task 2: Create and Initialize an RL Environment
# =====================================================
print("=" * 60)
print("TASK 2: Create and Initialize an RL Environment")
print("=" * 60)
 
env = gym.make("CartPole-v1")
 
# Reset the environment - returns initial observation + info dict
initial_observation, info = env.reset(seed=42)
 
print(f"Initial Observation: {initial_observation}")
print(f"Environment Info: {info}")
print(f"Environment Spec: {env.spec.id}\n")
```

```python

 
# =====================================================
# Task 3: Explore Observation and Action Spaces
# =====================================================
print("=" * 60)
print("TASK 3: Explore Observation and Action Spaces")
print("=" * 60)
 
print(f"Observation Space: {env.observation_space}")
print(f"Action Space: {env.action_space}")
print(f"Type of Observation Space: {type(env.observation_space)}")
print(f"Number of possible actions: {env.action_space.n}\n")
 
# Bonus context: what each observation value physically means
print("Observation breakdown (CartPole-v1):")
labels = ["Cart Position", "Cart Velocity", "Pole Angle", "Pole Angular Velocity"]
for label, low, high in zip(labels, env.observation_space.low, env.observation_space.high):
    print(f"  {label:25s} -> low: {low:>10.3f}, high: {high:>10.3f}")
print()
 
```

```python

# =====================================================
# Task 4: Execute a Random Agent
# =====================================================
print("=" * 60)
print("TASK 4: Execute a Random Agent")
print("=" * 60)
 
observation, info = env.reset(seed=42)
total_reward = 0
step_count = 0
terminated = False
truncated = False
 
print(f"{'Step':<6}{'Action':<8}{'Observation':<45}{'Reward':<8}{'Terminated':<12}{'Truncated'}")
print("-" * 95)
 
while not (terminated or truncated):
    action = env.action_space.sample()          # random action
    observation, reward, terminated, truncated, info = env.step(action)
 
    total_reward += reward
    step_count += 1
 
    obs_str = np.array2string(observation, precision=3, floatmode='fixed')
    print(f"{step_count:<6}{action:<8}{obs_str:<45}{reward:<8}{str(terminated):<12}{str(truncated)}")
 
print("-" * 95)
print(f"\nEpisode finished.")
print(f"Total number of steps: {step_count}")
print(f"Cumulative reward: {total_reward}")
 
env.close()
 
```
